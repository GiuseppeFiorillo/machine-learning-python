# Mini-progetto 1 — Registro dei risultati

Costruiamo un piccolo registro che usa dizionari, liste, funzioni e validazione. Ogni studente è associato a una lista di voti. Il programma aggiunge misure, calcola statistiche e stampa un report.

## Requisiti

1. Conservare i voti in un dizionario `nome -> lista di valori`.
2. Accettare soltanto voti interi tra 18 e 30.
3. Calcolare media, minimo e massimo di ogni studente.
4. Ordinare il report dalla media più alta alla più bassa.
5. Gestire input non validi senza corrompere il registro.

## Progettazione delle funzioni

Prima di scrivere il codice, dividiamo il problema in funzioni con responsabilità distinte:

| Funzione | Responsabilità |
| --- | --- |
| `validate_grade` | Controllare l'intervallo di voto |
| `add_grade` | Validare e aggiornare il registro |
| `summarize` | Calcolare statistiche per una lista |
| `build_report` | Costruire e ordinare righe di report |
| `print_report` | Presentare i risultati senza calcolarli|

<b>Ordine sicuro</b><br>

La validazione avviene prima delle modifiche del dizionario. In questo modo, se il dato è invalido, il registro resta nello stato precedente.
</div>

## Implementazione completa

```python
from __future__ import annotations

GradeBook = dict[str: list[int]]

def validate_grade(grade: int) -> int:
    if not 18 <= grade <= 30:
        raise ValueError("Il voto dev'essere compreso tra 18 e 30.")
    return grade

def add_grade(book: GradeBook, student_name: str, grade: int) -> None:
    clean_name = student_name.strip()
    if not clean_name:
        raise ValueError("Il nome non può essere vuoto.")
    
    validated = validate_grade(grade)
    book.setdefault(clean_name, []).append(validated)

def summarize(grades: list[int]) -> dict[str: float | int]:
    if not grades:
        raise valueError("La lista non può essere vuota.")
    
    return {
        "average": sum(grades) / len(grades),
        "minimum": min(grades),
        "maximum": max(grades),
        "count": len(grades),
    }

def build_report(book:GradeBook) -> list[dict[str: object]]:
    report = []
    for student, grades in book.items():
        stats = summarize(grades)
        report.append({"student": student, **stats})

    return sorted(
        report,
        key=lambda row: row["average"],
        reverse=True,
    )

def print_report(report:list[dict[str: object]]) -> None:
    print("\n=== Report ===")
    for position, row in enumerate(report, start=1):
        print(
            f"{position}. {row['student']}, "
            f"media={row['average']:.2f}, "
            f"min={row['minimum']}, "
            f"max={row['maximum']}, "
            f"n={row['count']}"
        )

def main() -> None:
    book: GradeBook = {
        "Amedeo": [28, 30, 27],
        "Onorato": [24, 26],
        "Rosa": [30, 29, 30],
    }

    try:
        add_grade(book, "Onorato", 28)
        add_grade(book, "Bruno", 31)
    except ValueError as error:
        print(f"Dato rifiutato: {error}")
    
    report = build_report(book)
    print_report(report)

if __name__ == "__main__":
    main()
```

## Cosa osservare?

* `setdefault` crea la lista soltanto per un nuovo studente;
* `summarize` è una funzione quasi pura: legge la lista e restituisce un nuovo dizionario;
* l'espressione `{"studente": student, **stats}` combina un campo con un dizionario esistente;
* `sorted` usa una lambda per ordinare in base alla media;
* la validazione avviene prima della modifica, quindi un valore invalido non altera il registro.
