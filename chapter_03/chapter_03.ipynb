# Capitolo 3 — Strutture dati e funzioni

## Perché servono le strutture dati?

Una variabile singola è adatta a rappresentare un solo valore.

```python
temperature_1 = 18.2
temperature_2 = 19.0
temperature_3 = 20.1
```

Creare una variabile per ogni misura, però, non scala. Una struttura dati riunisce più valori sotto un unico nome:

```python
temperatures = [18.2, 19.0, 20.1]
print(temperatures)
```

### Mutabilità e identità degli oggetti

Una struttura è mutabile quando può cambiare dopo la creazione. Liste, dizionari e set sono mutabili; tuple, stringhe, interi e float sono immutabili. La mutabilità influenza il comportamento delle assegnazioni e delle funzioni.

```python
values = [10, 20, 30]
print(id(values))
values.append(40)
print(values)
print(id(values)) # lo stesso oggetto è stato modificato
```

In Python una variabile non contiene direttamente il valore: è un nome associato a un oggetto. Due nomi possono quindi riferirsi allo stesso oggetto mutabile.

## Liste

Una lista è una sequenza ordinata e mutabile. Può contenere zero o più elementi, anche di tipi differenti. Nei progetti di analisi dati conviene tuttavia mantenere elementi omogenei quando rappresentano la stessa grandezza.

```python
empty = []
features = ["temperature", "humidity", "pressure"]
scores = [0.72, 0.81, 0.78]
mixed = ["experiment_1", 50, True]

print(features[0]) # primo elemento
print(features[-1]) # ultimo elemento
```

Gli indici partono da zero. Con un indice negativo si conta dalla fine:

| Espressione | Elemento selezionato |
| --- | --- |
| `features[0]` | Primo elemento |
| `features[1]` | Secondo elemento |
| `features[-1]` | Ultimo elemento |
| `features[-2]` | Penultimo elemento |

### Modifica degli elementi

```python
metrics = [0.80, 0.74, 0.69]
metrics[1] = 0.77
metrics[-1] = 0.71
print(metrics)
```

La modifica avviene sullo stesso oggetto. È possibile anche sostituire una porzione della lista con uno slicing:

```python
values = [1, 2, 3, 4, 5]
values[1:4] = [20, 30, 40]
print(values)
```

### Metodi fondamentali delle liste

| Metodo | Effetto |
| --- | --- |
| `append(x)` | Aggiunge `x` in fondo |
| `extend(iterabile)` | Aggiunge tutti gli elementi di un iterabile |
| `insert(i, x)` | Inserisce `x` nella posizione `i` |
| `remove(x)` | Elimina la prima occorrenza di `x` |
| `pop(i)` | Rimuove e restituisce l’elemento in posizione `i` |
| `clear()` | Rimuove tutti gli elementi |
| `index(x)` | Restituisce l’indice della prima occorrenza |
| `count(x)` | Conta quante volte compare `x` |
| `sort()` | Ordina la lista sul posto |
| `reverse()` | Inverte la lista sul posto |
| `copy()` | Crea una copia superficiale |

```python
features = ["temperature", "humidity"]
features.append("pressure")
features.extend(["wind_speed", "rain_rate"])
features.insert(1, "station_id")

removed = features.pop()
print(features)
print("Rimosso: ", removed)
```

**`append` ed `extend` non sono equivalenti**

```python
a = [1, 2]
a.append([3, 4])
print(a)

b = [1, 2]
b. extend([3, 4])
print(b)
```

`append` aggiunge un solo oggetto, anche quando tale oggetto è una lista. `extend` scorre invece l'iterabile e aggiunge i suoi elementi uno alla volta.

### Slicing

Lo slicing seleziona una sottosequenza secondo la forma sequenza `[start:stop:step]`. L'estremo `stop` non è incluso.

```python
values = [10, 20, 30, 40, 50, 60]

print(values[1:4]) # [20, 30, 40]
print(values[:3]) # [10, 20, 30]
print(values[3:]) # [40, 50, 60]
print(values[::2]) # [10, 30, 50]
print(values[::-1]) # lista invertita
```

| Espressione | Interpretazione |
| -- | -- |
| `values[a:b]` | Da `a` incluso a `b` escluso |
| `values[:b]` | Dall'inizio a `b` escluso |
| `values[a:]` | Da a fino alla fine |
| `values[::k]` | Un elemento ogni `k` |
| `values[::-1]` | Copia in ordine inverso |

```python
metrics = [0.71, 0.78, 0.83]

for metric in metrics:
    print(metric)
```

### Aliasing, copia superficiale e copia profonda

L'assegnazione di una lista a un altro nome non crea una copia:

```python
original = [1, 2, 3]
alias = original
alias.append(4)

print(original)
print(alias is original)
```

Per una lista piatta si può creare una copia superficiale con `copy()` o con uno slicing completo:

```python
original = [1, 2, 3]
copy_a = original.copy()
copy_b = original[:]

copy_a.append(4)
print(original) # [1, 2, 3]
```

Con strutture annidate, una copia superficiale duplica solo il contenitore esterno. Gli oggetti interni restano condivisi:

```python
rows = [[1, 2], [3, 4]]
shallow = rows.copy()
shallow[0].append(99)

print(rows)
```

Quando serve duplicare ricorsivamente tutta la struttura, usa `copy.deepcopy`:

```python
from copy import deepcopy

rows = [[1, 2], [3, 4]]
independent = deepcopy(rows)
independent[0].append(99)

print(rows) # [[1, 2], [3, 4]]
print(independent) # [[1, 2, 99], [3, 4]]
```

## Tuple

Una tupla è una sequenza ordinata e immutabile. Si usa quando un gruppo di valori deve restare concettualmente stabile oppure quando si vuole rappresentare un record breve e posizionale.

### Creazione e unpacking

```python
image_size = (640, 480)
coordinates = (12.5, 9.8)
empty = ()
single = (42,) # la virgola crea una tupla con un elemento

width, height = image_size
print(width)
print(height)
```

L'unpacking assegna gli elementi della seguenza a più variabili. Il numero di nomi deve coincidere con il unmero di elementi, salvo l'uso di `*`:

```python
first, *middle, last = [10, 20, 30, 40, 50]
print(first)
print(middle)
print(last)
```

### Immutabilità

```python
shape = (32, 64)

#shape[0] = 128 # Type Error
shape = (128, 64) # il nome viene associato a una nuova tupla
```

Immutabile non significa che ogni oggetto contenuto sia immutabile a sua volta. Una tupla può contenere una lista, e quella lista può cambiare:

```python
configuration = ("experiment_1", [32, 64])
configuration[1].append(128)
print(configuration)
```

### Quando scegliere una tupla?
* dimensioni di un'immagine, coordinate o intervalli;
* valori restituiti insieme da una funzione;
* chiavi composte di un dizionario, purché tutti gli elementi siano hashable;
* record brevi che non devono essere modificati accidentalmente;
* dati che vengono passati come configurazione fissa.

```python
def min_max(values):
    return min(values), max(values)

minimum, maximum = min_max([8, 3, 12, 5])
print(minimum, maximum)
```

## Dizionari

Un dizionario associa chiavi uniche a valori. È una struttura fondamentale per record, configurazioni, metriche e metadati. Le chiavi devono essere oggetti hashable, come stringhe, numeri o tuple immutabili.

### Creazione e lettura

```python
model_config = {
    "name": "baseline",
    "learning_rate": 0.001,
    "batch_size": 32,
    "shuffle": True,
}

print(model_config["learning_rate"])
print(model_config.get("epochs"))
print(model_config.get("epochs", 10))
```

L'accesso con parentesi quadre genera `KeyError` se la chiave non esiste. Il metodo `get` restituisce invece `None` o un valore di default.

### Inserimento, aggiornamento e rimozione

```python
config = {"epochs": 20, "batch_size": 32}

config["learning_rate"] = 0.001 # nuova coppia
config["epochs"] = 30           # aggiornamento
config.update({"optimizer": "adam", "weight_decay": 0.0001})

removed = config.pop("batch_size")
print(removed)
print(config)
```

Altri strumenti utili sono `del`, `popitem` e `clear`. Usa `pop` quando ti serve anche il valore rimosso.

### Iterazione su chiavi, valori e coppie

```python
metrics = {
    "accuracy": 0.91,
    "precision": 0.88,
    "recall": 0.84,
}

for name in metrics:
    print(name)

for value in metrics.values():
    print(value)

for name, value in metrics.items():
    print(f"{name}: {value:.2f}")
```

### Appartenenza e metodi sicuri

```python
config = {"epochs": 30, "batch_size": 32}

if "learning_rate" not in config:
    config["learning_rate"] = 0.001

config.setdefault("optimizer", "adam")
print(config)
```

L'operatore `in` su un dizionario verifica le chiavi, non i valori. `setdefault` inserisce il default solo quando la chiave è assente e restituisce il valore associato.

### Dizionari annidati e record

```python
experiment = {
    "id": "exp_001",
    "config": {
        "learning_rate": 0.001,
        "batch_size": 32
    },
    "metrics": {
        "train_loss": [0.82, 0.61, 0.49],
        "val_loss": [0.88, 0.70, 0.63],
    },
}

last_val_loss = experiment["metrics"]["val_loss"][-1]
print(last_val_loss)
```

## Set
Un set è una collezione mutabile di elementi unici. È ottimizzato per controllare l'appartenenza, eliminare duplicati e calcolare operazioni insiemistiche.

### Creazione e modifica

```python
labels = {"cat", "dog", "bird"}
empty_set = set() # {} creerebbe un dizionario vuoto

labels.add("horse")
labels.discard("bird")
labels.remove("dog")
print(labels)
```

`discard` non genera errori se l'elemento manca; `remove` genera `KeyError`.

### Operazioni insiemistiche

```python
train_labels = {"cat", "dog", "bird"}
test_labels = {"cat", "dog", "horse"}

print(train_labels | test_labels) # unione
print(train_labels & test_labels) # intersezione
print(train_labels - test_labels) # differenza
print(train_labels ^ test_labels) # differenza simmetrica
```

| Operazione | Metodo equivalente | Significato |
| --- | --- | --- |
| `a \| b` | `a.union(b)` | Elementi presenti in almeno uno |
| `a & b` | `a.intersection(b)` | Elementi presenti in entrambi |
| `a - b` | `a.difference(b)` | Elementi di `a` non presenti in `b` |
| `a ^ b` | `a.symmetric_difference(b)` | Elementi presenti in uno solo |
| `a <= b` | `a.issubset(b)` | `a` è sottoinsieme di `b` |
| `a >= b` | `a.issuperset(b)` | `a` contiene `b` |

### Rimozione dei duplicati e membership

```python
raw_ids = [3, 1, 3, 2, 1, 4]
unique_ids = set(raw_ids)
print(unique_ids)

allowed = {"train", "validation", "test"}
split = "validation"

if split in allowed:
    print("Split valido")
```

## Comprehension
Le comprehension costruiscono collezioni applicando un'espresione agli elementi di un iterabile. Sono compatte e idiomatiche, ma devono restare leggibili.

### List comprehension

```python
squares = []
for number in range(6):
    squares.append(number ** 2)

# Forma equivalente
squares = [number ** 2 for number in range(6)]
print(squares)
```

La struttura generale è:

$[\text{espressione for elemento in iterabile}]$

### Filtri e condizioni

```python
values = [-3, 5, 0, 8, -1, 4]
positive = [value for value in values if value > 0]
labels = ["positive" if value > 0 else "not_positive" for value in values]
```

Il filtro finale decide quali elementi includere. L'operatore condizionale dentro l'espressione produce invece un valore per ogni elemento.

### Dictionary e set comprehension

```python
features = ["temperature", "humidity", "pressure"]
feature_to_index = {name: index for index, name in enumerate(features)}

words = ["Train", "train", "TEST", "test"]
normalized = {word.lower() for word in words}

print(feature_to_index)
print(normalized)
```

### Comprehension annidate

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
]

flattened = [value for row in matrix for value in row]
print(flattened)
```

Leggi la comprehension nello stesso ordine dei cicli equivalenti:

```python
flattened = []
for row in matrix:
    for value in row:
        flattened.append(value)
```

## Funzioni
Una funzione raccoglie una procedura sotto un nome. Riduce duplicazioni, separa responsabilità e consente di testare parti del programma in modo indipendente.

### Definizione e chiamata

```python
def greet():
    print("Benvenuto nel corso!")

greet()
```

La parola chiave `def` introduce la definizione. Il corpo è indentato. La funzione viene eseguita solo quando è chiamata. 

### Parametri e argomenti

```python
def describe_experiment(name, epochs):
    print(f"Esperimento: {name}")
    print(f"Epoche: {epochs}")

describe_experiment("baseline", 20)
```

I **parametri** sono i nomi nella definizione; gli **argomenti** sono i valori forniti nella chiamata.

### Valore di ritorno

```python
def mean(values):
    if not values:
        return None
    return sum(values) / len(values)

result = mean([0.8, 0.9, 0.7])
print(f"{result:.1f}")
```

`return` termina la funzione e restituisce un oggetto al chiamante. Se una funzione non incontra alcun `return`, restituisce implicitamente `None`.

#### Restituire più valori

```python
def summarize(values):
    return min(values), max(values), sum(values) / len(values)

minimum, maximum, average = summarize([3, 8, 4, 10])
```

Tecnicamente la funzione restituisce una sola tupla, poi scomposta mediante unpacking.

### Argomenti posizionali, keyword e default

```python
def train_message(model_name, epochs=10, verbose=True):
    if verbose:
        print(f"Training {model_name} per {epochs} epoche")

train_message("baseline")
train_message("transformer", 30)
train_message(model_name="cnn", verbose=False, epochs=20)
```

I parametri con default devono seguire quelli senza default. Gli argomenti keyword rendono le chiamate più chiare e consentono di cambiare l'ordine.

### Numero variabile di argomenti
`*args` raccoglie argomenti posizionali aggiuntivi in una tupla; `**kwargs` raccoglie argomenti keywords aggiuntivi in un dizionario.

```python
def average(*values):
    if not values:
        raise ValueError("Serve almeno un valore")
    return sum(values) / len(values)

def show_config(**config):
    for key, value in config.items():
        print(f"{key}: {value}")

print(average(0.8, 0.9, 0.7))
show_config(epochs=20, batch_size=32, optimizer="adam")
```

**Unpacking nella chiamata**

```python
values = [3, 8, 4]
print(max(*values))

config = {"model_name": "cnn", "epochs": 20}
train_message(**config)
```

L'asterisco non appartiene solo alla definizione: nella chiamata espande una sequenza in argomenti posizionali o un dizionario in argomenti keyword.

### Funzioni pure ed effetti collaterali
Una funzione pura dipende solo dagli argomenti e restituisce il risultato senza modificare lo stato esterno. Le funzioni pure sono più facili da testare e riutilizzare.


```python
def normalized(values):
    total = sum(values)
    if total == 0:
        return [0.0 for _ in values]
    return [value / total for value in values]

raw = [2.0, 3.0, 5.0]
result = normalized(raw)
print(raw) # invariata
print(result)
```

Stampare, scrivere file, modificare una lista ricevuta o aggiornare una variabile globale sono effetti collaterali. Non sono sempre sbagliati, ma devono essere intenzionali e riconoscibili.

## Scope e passaggio degli oggetti

### Scope locale e globale

```python
learning_rate = 0.001 # scope globale

def show_rate():
    message = "Learning rate" # scope locale
    print(message, learning_rate)

show_rate()
# print(message) # NameError
```

Python cerca i nomi secondo la regola LEGB: Local, Enclosing, Global, Built-in. In genere è meglio passare i dati come argomenti invece di dipendere da variabili globali.

### `global` e `nonlocal`

`global` permette di riassegnare un nome globale; `nonlocal` permette di riassegnare un nome nello scope di una funzione esterna. Sono strumenti validi, ma un uso frequente può indicare una progettazione poco chiara.

```python
counter = 0

def increment():
    global counter
    counter += 1
```

### Python passa riferimenti a oggetti

Python non usa semplicemente il modello "per valore" o "per riferimento" tipico di altri linguaggi. La funzione riceve un nuovo nome associato allo stesso oggetto passato dal chiamante.

```python
def append_score(scores, value):
    scores.append(value)

metrics = [0.80, 0.82]
append_score(metrics, 0.85)
print(metrics) # modificata
```

Con un oggetto immutabile, una riassegnazione locale non modifica il nome esterno:

```python
def increment(value):
    value += 1
    return value

count = 10
new_count = increment(count)
print(count)     # 10
print(new_count) # 11
```

## Funzioni come oggetti e lambda

In Python le funzioni sono oggetti di prima classe: possono essere assegnate a variabili, inserite in collezioni, passate come argomenti e restituite da altre funzioni.

### Passare una funzione

```python
def square(value):
    return value ** 2

def apply_operation(values, operation):
    return [operation(value) for value in values]

result = apply_operation([1, 2, 3], square)
print(result)
```

### Espressioni lambda

Una lambda crea una piccola funzione anonima composta da una sola espressione:

```python
square = lambda value: value ** 2
print(square(5))
```

La forma equivalente con `def` è spesso più leggibile quando la logica richiede un nome o richiede più righe.

### Ordinamento con `key`

```python
experiments = [
    {"name": "cnn", "score": 0.88},
    {"name": "svm", "score": 0.82},
    {"name": "transformer", "score": 0.91},
]

ranked = sorted(
    experiments,
    key = lambda experiment: experiment["score"],
    reverse = True
)

for experiment in ranked:
    print(experiment["name"], experiment["score"])
```

### `map` e `filer`

`map()` applica una funzione a ogni elemento di una lista e restituisce un nuovo oggetto con i risultati trasformati; `filter()` filtra gli elementi di una lista mantenendo solo quelli per cui la funzione restituisce `True`.

```python
values = [1, 2, 3, 4]

squares = list(map(lambda value: value ** 2, values))
even = list(filter(lambda value: value % 2 == 0, values))
```

In Python una comprehension è spesso più leggibile.

```python
squares = [value ** 2 for value in values]
even = [value for value in values if value % 2 == 0]
```

### Type hint

```python
from collections.abc import Sequence

def mean(values: Sequence[float]) -> float:
    if not values:
        raise ValueError("La sequenza non può essere vuota")
    return sum(values) / len(values)
```

Le annotazioni non impediscono automaticamente di passare un tipo errato. Strumenti come linting e type checker possono analizzarle prima dell'esecuzione.

## Gestione delle eccezioni

Un programma robusto distingue gli errori prevedibili dai bug. Le eccezioni permettono di interrompere il flusso normale e trasferire il controllo a un blocco capace di gestire il problema.

### Errori di sintassi, runtime e logica

| Categoria | Descrizione | Esempio |
| --- | --- | --- |
| Sintassi | Il codice non rispetta la grammatica di Python | Parentesi mancante |
| Runtime | Il codice parte ma un’operazione fallisce | `int("abc")` |
| Logica | Il programma termina ma produce un risultato errato | Media divisa per $n - 1$ |

### `try` ed `except`

```python
raw_value = input("Inserisci un numero: ")

try:
    value = float(raw_value)
except ValueError:
    print("Il testo inserito non è un numero valido")
else:
    print(f"Valore convertito: {value}")
```

Cattura l'eccezione più specifica possibile. Un `except Exception` troppo ampio può nascondere bug che dovrebbero essere corretti.

### `else` e `finally`

```python
file = None

try:
    file = open("measurements.txt", encoding="utf-8")
    content = file.read()
except FileNotFoundError:
    print("File non trovato")
else:
    print(f"Caratteri letti: {len(content)}")
finally:
    if file is not None:
        file.close()
```

`else` viene eseguito quando non si verificano eccezioni. `finally` viene eseguito comunque ed è utile per rilasciare risorse. Per i file useremo normalmente il context manager `with`.

### Sollevare eccezioni con `raise`

```python
def validate_probability(value: float) -> float:
    if not 0.0 <= value <= 1.0:
        raise ValueError("La probabilità dev'essere tra 0 e 1")
    return value
```

Sollevare un'eccezione rende esplicito il contratto della funzione e impedisce che dati non validi proseguano nella pipeline.

### Gestire l'eccezione al livello giusto

```python
def parse_epoch(raw: str) -> int:
    epoch = int(raw)
    if epoch <= 0:
        raise ValueError("Il numero di epoche dev'essere positivo")
    return epoch

def main() -> None:
    raw = input("Epoche: ")
    try:
        epoch = parse_epoch(raw)
    except ValueError as error:
        print(f"Configurazione non valida: {error}")
        return

    print(f"Avvio per {epoch} epoche")
```
